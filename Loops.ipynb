# while loop

```python
counter = 0

while counter < 10:
    counter += 1
    print(counter)
```

```python
stock_portfolio = 800000
year_counter = 0

while stock_portfolio < 1000000:
    investment_income = stock_portfolio * .05
    stock_portfolio += investment_income
    year_counter += 1
    print(f"Year: {year_counter} " + f"My balance is {stock_portfolio}")
```

```python
bank_balance = 5000
monthly_counter = 0

while bank_balance > 0:
    spending = 1000
    bank_balance -= spending
    monthly_counter += 1
    
    print(f"Month: {monthly_counter} My balance is {bank_balance}")
```

```python
current_inventory = 686
monthly_sales = 84
month = 1

while current_inventory > 0:
    current_inventory -= monthly_sales
    month += 1
    print(f"At the end of month {month}, we have {current_inventory} pairs of inventory'")
```

# try and except

```python
price_list = [5.99, None, 19.99, 24.99, 0, '74.99', 99.99]

#loop to calculate how many of each item I can buy for 50 dollars
for price in price_list:
affordable_quantity = 50//price # My budget is 50 dollars
print(f"I can buy {affordable_quantity} of these.")
```

```python
price_list = [5.99, None, 19.99, 24.99, 0, '74.99', 99.99]

for price in price_list:
    try:
        affordable_quantity = 50//price
        print(f"I can buy {affordable_quantity} of these.")
    except:
        print("The price seems to be missing")
```

```python
price_list = [5.99, None, 19.99, 24.99, 0, '74.99', 99.99]

for price in price_list:
    try:
        affordable_quantity = 50//price
        print(f"I can buy {affordable_quantity} of these.")
    except ZeroDivisionError:
        print("The price seems to be missing")
    except:
        print("That's not a number")
```

# Loop control example

```python
word = "Maven"
new_word = ""

for letter in word:
    if letter in "aeiou":
        continue
    new_word += letter
    
new_word
```

```python
word = "Maven"
new_word = ""

for letter in  word:
    if letter in "aeiou":
        pass
    new_word += letter

new_word
```

```python
word = "Maven"
new_word = ""

for letter in  word:
    if letter == "v":
        break
    new_word += letter

new_word
```

```python
ratings = ['1 star', '5 star', '3 star', 'star 5', '4 star']

for  rating in ratings:
    try:
        print(int(rating[0]))
    except ValueError:
        print("incorrect data")
```

```python
number = 28

if number > 1:
    for i in range(2,number):
        if number % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
else:
    print("Not a prime number")
```

```python
number = 29
it_prime = True

if number > 1:
    for i in range(2, number):
        if number % i == 0:
            it_prime = False
            break
print(it_prime)
```

```python
items = ["apple", "banana", "orange", "apple", "mango"]

unique_item = set()

for item in items:
    if item in unique_item:
        print("Duplicate item: ", item)
        break
    unique_item.add(item)
```
